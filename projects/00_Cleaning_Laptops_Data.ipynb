# Cleaning Laptops Data

CSV file containing information about 1,300 laptop computers. The first five rows of the CSV file are shown below:

```python
# import libraries
import pandas as pd
```

```python
# data set
laptops = pd.read_csv('laptops.csv', encoding='latin-1')
laptops.info()
```

We can see that every column is represented as the object type, indicating that they are represented by strings, not numbers. Also, one of the columns, Operating System Version, has null values.

The column labels have a variety of upper and lowercase letters, as well as spaces and parentheses, which will make them harder to work with and read. One noticeable issue is that the `" Storage"` column name has a space in front of it.

```python
# old column
old_columns = laptops.columns
print('Old Columns: ', old_columns)

# remove any whitespace of each column
new_columns = []
for col in old_columns:
    new_columns.append(col.strip())

print('New Columns: ', new_columns)   
```

However, the column labels still have a variety of upper and lowercase letters, as well as parentheses, which will make them harder to work with and read. Let's finish cleaning our column labels by:

- Replacing spaces with underscores.
- Removing special characters.
- Making all labels lowercase.
- Shortening any long column names.

```python
# function to clean columns labels
def clean_col(col):
    col = col.strip()
    col = col.replace('(','')
    col = col.replace(')','')
    col = col.lower()
    col = col.replace('Operating System', 'os')
    col = col.replace(' ','_')
    return col
```

```python
# cleaning column labels
new_columns = []
for col in laptops.columns:
    new_label = clean_col(col)
    new_columns.append(new_label)
    
print('New Columns: ', new_columns)    
```

```python
laptops.columns = new_columns
```

We observed earlier that all 13 columns have the object dtype, meaning they're stored as strings. Let's look at the first few rows of some of our columns:

```python
# first rows
print(laptops.iloc[:5,2:5])
```

We want to explore data:

```python
# explore the screen_size
print(laptops['screen_size'].dtype)
print(laptops['screen_size'].unique())
```

All values in this column follow the same pattern, and there are no special cases. We will need to convert the column to a float type.

```python
laptops['screen_size'] = laptops['screen_size'].str.replace('"','')
print(laptops['screen_size'].unique())
```

```python
laptops['screen_size'] = laptops['screen_size'].astype(float)
print(laptops['screen_size'].unique())
```

```python
# explore ram column
print(laptops['ram'].dtype)
print(laptops['ram'].unique())
```

We identified a clear pattern in the `ram` column - all values are integers and include the character `GB` at the end of the string

```python
laptops['ram'] = laptops['ram'].str.replace('GB','').astype(int)
print(laptops['ram'].unique())
```

We have converted some columns to numeric dtypes, and we need to rename the column because it is usefull if the non-digit values contain information that help us to understand the data

```python
# rename screen_size
laptops.rename({'screen_size':'screen_size_inches'}, axis=1, inplace=True)

# rename ram
laptops.rename({'ram':'ram_gb'}, axis=1, inplace=True)

# describe
print(laptops['screen_size_inches'].describe())
print('\n')
print(laptops['ram_gb'].describe())
```

If your data has been scraped from a webpage or if there was manual data entry involved at some point, you may end up with inconsistent values. Let's look at an example from our os column:

```python
# values operating system column
laptops['operating_system'].value_counts()
```

We can see that there are two variations of the Apple operating system — `macOS` — in our dataset: `Mac OS` and `macOS`

```python
# corrected values in operating system column
mapping_dict = {
    'Android': 'Android',
    'Chrome OS': 'Chrome OS',
    'Linux': 'Linux',
    'Mac OS': 'macOS',
    'No OS': 'No OS',
    'Windows': 'Windows',
    'macOS': 'macOS'
}
# columns correction
laptops['operating_system'] = laptops['operating_system'].map(mapping_dict)

# corrected values operating system column
laptops['operating_system'].value_counts()
```

Dropping Missing Values

```python
# cheking null values
laptops.isnull().sum()
```
