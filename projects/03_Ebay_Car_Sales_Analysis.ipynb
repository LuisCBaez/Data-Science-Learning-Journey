## 6. Dealing with Incorrect Registration Year Data

One thing that stands out from the exploration we did is that the `registration_year` column contains some odd values:

- The minimum value is 1000, before cars were invented

- The maximum value is 9999, many years into the future

Because a car cannot be first registered after the listing was seen, any vehicle with a registration year above 2016 is definitely inaccurate. Determining the earliest valid year is more difficult. Realistically, it could be somewhere in the first few decades of the 1900s.

```python
# deleting values outside the 1900 - 2016 interval
autos = autos[autos['registration_year'].between(1900, 2016)]

# calculating the distribution
autos['registration_year'].value_counts(normalize=True)
```

## 7. Exploring Price by Brand

When working with data on cars it's natural to explore variations across different car brands. We can use aggregation to understand the brand column.

```python
# identifying the unique values we want to aggregate by
autos['brand'].value_counts().head(20)
```

```python
# assigning the index of the top 20 car brands
top_20_brands = autos['brand'].value_counts().head(20).index
top_20_brands
```

```python
# creating an empty dictionary to store our aggregate data
top_20_prices = {}

# loop over the top 20 brands
for each_brand in top_20_brands:
    top_autos = autos[autos['brand'] == each_brand]
    mean_price = int(top_autos['price'].mean())                   
    top_20_prices[each_brand] = mean_price

# printing the dictionary
print(top_20_prices)
```

We observed that in the top 20 brands, there's a distinct price gap:

- Audi, BMW and Mercedes Benz are more expensive
- Renault and Fiat are less expensive
- Volkswagen is in between


## 8. Storing Aggregate Data in a DataFrame



For the top 20 brands, we use aggregation to understand the average mileage for those cars and if there's any visible link with mean price.

First, we create an empty dictionary named `top_20_mileage` to save the mean mileage per each brand.

```python
# creating an empty dictionary to store our aggregate data
top_20_mileage = {}

# loop over the top 20 brands
for each_brand in top_20_brands:
    top_autos = autos[autos['brand'] == each_brand]
    mean_mileage = int(top_autos['odometer_km'].mean())
    top_20_mileage[each_brand] = mean_mileage
    
# printing the dictionary
print(top_20_mileage)
```

We'll combine the data from series objects into a single dataframe (with a shared index, `top_20_brands`) and display the dataframe directly.

```python
# converting dictionaries to series objects
top_20_prices_series = pd.Series(top_20_prices)
top_20_mileage_series = pd.Series(top_20_mileage)

# creating a dataframe from the series objects
top_brands = pd.DataFrame(top_20_prices_series, columns=['mean_price'])

# assigning the series object in the dataframe
top_brands['mean_mileage'] = top_20_mileage_series

# printing the dataframe
top_brands
```

```python
# analizing the dataframe
print(top_brands.sort_values('mean_price', ascending=False))
print('\n')
print(top_brands.sort_values('mean_mileage', ascending=False))
```

As can be seen in both tables, although `audi`, `mercedes_benz` and `bmw` are the most expensives cars, these brands also have the highest mean mileage.

We will identify categorical data that uses German words, translate them and map the values to their English counterparts.

```python
# identifying patterns and special cases
print('Column - vehicle_type')
print(autos['vehicle_type'].unique())
print('\n')
print('Column - gearbox')
print(autos['gearbox'].unique())
print('\n')
print('Column- fuel_type')
print(autos['fuel_type'].unique())
print('\n')
print('Column - unrepaired_damage')
print(autos['unrepaired_damage'].unique())
print('\n')
print(autos.isnull().sum())
```

We identified German words in four columns in our dataset (`vehicle_type`, `gearbox`, `fuel_type` and `unrepaired_damage`). In addition, these columns and `model` column contain null (or NaN) values that can be removed.

```python
# removing rows with null (or NaN) values
autos = autos.dropna(axis=0)
```

```python
# checking null values
print(autos.isnull().sum())
```

```python
# creating a dictionaries with the corrections
vehicle_type_corrections = {'bus': 'bus', 'limousine': 'limousine',
                            'kleinwagen': 'small_car', 'kombi': 'combi',
                            'coupe': 'coupe', 'suv': 'suv',
                            'cabrio': 'convertible', 'andere': 'other'}

gearbox_corrections = {'manuell': 'manually', 'automatik': 'automatic'}

fuel_type_corrections = {'lpg': 'lpg', 'benzin': 'petrol',
                         'diesel': 'diesel', 'cng': 'cng',
                         'hybrid': 'hybrid', 'elektro':'electro',
                         'andere':'other'}

unrepaired_damage_corrections = {'nein':'no', 'ja': 'yes'}

# renaming values
autos['vehicle_type'] = autos['vehicle_type'].map(vehicle_type_corrections)

autos['gearbox'] = autos['gearbox'].map(gearbox_corrections)

autos['fuel_type'] = autos['fuel_type'].map(fuel_type_corrections)

autos['unrepaired_damage'] = autos['unrepaired_damage'].map(unrepaired_damage_corrections)
```

```python
autos.head()
```

We will convert the dates to be uniform numeric data, so "2016-03-26 17:47:46" becomes the integer "20160326".

```python
# converting dates strings into integers
autos['date_crawled'] = autos['date_crawled'].str[:10].str.replace('-', '').astype(int)

autos['ad_created'] = autos['ad_created'].str[:10].str.replace('-', '').astype(int)

autos['last_seen'] = autos['last_seen'].str[:10].str.replace('-', '').astype(int)

# printing dates
print(autos[['date_crawled', 'ad_created', 'last_seen']])
```

### 9.2. Analysis:

We will do:

- Find the most common `brand/model` combinations   
- Split the `odometer_km` into groups and use aggregation to see the average prices based on the mileage   
- Answer the question: How much cheaper are cars with damage than their non-damaged counterparts?

```python
# finding the most common combinations
brand_model = autos.groupby(['brand','model'])['name'].count().sort_values(ascending=False).head(10)
brand_model    
```

According to the results, Volkswagen is the most popular brand with its models `golf`, `polo` and `passat` which sum `4,906` cars in total on the 10 most common brand/model.

The most common brand/model is `Volkswagen - Golf` with 2,749 cars, followed by `BMW - 3er` and `Volkswagen - Polo` with 2,083 and 1,133 respectively.

```python
autos['odometer_km'].describe()
```

```python
# dataframe to analyze
mileage_price = autos[['odometer_km', 'price']]

# creating an dictionary to store our mean price
mieleage_intervals_km = {'0 - 30,000 km': 0, '30,000 - 60,000 km': 0, '60,000 - 90,000 km': 0, '90,000 - 120,000 km': 0, '120,000 - 150,000 km': 0}

# loop over the intervals
for interval in mieleage_intervals_km:    
    if interval == '0 - 30,000 km':
        km = mileage_price[mileage_price['odometer_km'] <= 30000]  
        mean_price = int(km['price'].mean())
        mieleage_intervals_km[interval] = mean_price  
        
    elif interval == '30,000 - 60,000 km':
        km = mileage_price[(mileage_price['odometer_km'] > 30000) & (mileage_price['odometer_km'] <= 60000)]  
        mean_price = int(km['price'].mean())
        mieleage_intervals_km[interval] = mean_price  
                            
    elif interval == '60,000 - 90,000 km':
        km = mileage_price[(mileage_price['odometer_km'] > 60000) & (mileage_price['odometer_km'] <= 90000)]  
        mean_price = int(km['price'].mean())
        mieleage_intervals_km[interval] = mean_price  
                            
    elif interval == '90,000 - 120,000 km':
        km = mileage_price[(mileage_price['odometer_km'] > 90000) & (mileage_price['odometer_km'] <= 120000)]  
        mean_price = int(km['price'].mean())
        mieleage_intervals_km[interval] = mean_price  
                            
    elif interval == '120,000 - 150,000 km':
        km = mileage_price[mileage_price['odometer_km'] > 120000]  
        mean_price = int(km['price'].mean())
        mieleage_intervals_km[interval] = mean_price
        
# converting dictionaries to series objects
mieleage_series = pd.Series(mieleage_intervals_km)

# creating a dataframe from the series objects
mieleage_data = pd.DataFrame(mieleage_series, columns=['mean_price'])
mieleage_data
```

As we can see in the dataframe, the mean price with respect to the mileage are indirectly proportional (The mean price decreases `$4,000` each `30,000 km`). 

```python
# assigning the index of the brands
brands_index = autos['brand'].value_counts().index

# creating an empty dictionary to store our aggregate data
damage_no = {}
damage_yes = {}

# loop over the brands to calculate the mean price of non damage and damage cars
for each_brand in brands_index:
    brand_bool_no = autos[(autos['brand'] == each_brand) & (autos['unrepaired_damage'] == 'no')]
    brand_bool_yes = autos[(autos['brand'] == each_brand) & (autos['unrepaired_damage'] == 'yes')]
    
    mean_price_no = int(brand_bool_no['price'].mean())
    mean_price_yes = int(brand_bool_yes['price'].mean())
 
    damage_no[each_brand] = mean_price_no
    damage_yes[each_brand] = mean_price_yes

# converting dictionaries to series objects
damage_no_prices = pd.Series(damage_no)
damage_yes_prices = pd.Series(damage_yes)

# creating a dataframe from the series objects
brands_damage = pd.DataFrame(damage_no_prices, columns=['mean_price_non_damage'])

# assigning the series object in the dataframe
brands_damage['mean_price_damage'] = damage_yes_prices

# calculating the difference between prices
brands_damage['price_difference'] = brands_damage['mean_price_non_damage'] - brands_damage['mean_price_damage']

# printing the dataframe
brands_damage
```

```python
# descriptive statistics
brands_damage['price_difference'].describe()
```

As a result of this analysis, Every brand has the highest mean price in cars with non damage. Besides, the average price difference between non damage and damage cars is around `$4,647`.
