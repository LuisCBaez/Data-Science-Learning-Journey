```python
class Employee:
    def __init__(self, first, last, pay):    # self. ---- first, last, pay are attributes of the Class
        self.first = first
        self.last = last
        self.pay = pay
        self.email = first + '.' + last + '@company.com'
              
```

## Tutorial 2: Class Variables

- Remember that methods automatically take in the instance with self.

```python
class Employee:
    def __init__(self, first, last, pay):
        # instances
        self.first = first
        self.last = last
        self.pay = pay
        self.email = first + '.' + last + '@company.com'
    
    # method
    def fullname(self):
        return '{} {}'.format(self.first, self.last)
    
    def apply_raise(self):
        self.pay = int(self.pay * 1.04) # raise of 4%
    
```

```python
emp_1 = Employee('Corey', 'Schafer', 50000)
emp_2 = Employee('Test', 'User', 60000)

print(emp_1.pay)
emp_1.apply_raise()
print(emp_1.pay)
```

## Tutorial 3: Classmethods and Staticmethods

### 3.1 Classmethods

```python
class Employee:
    
    num_of_emps = 0
    raise_amount = 1.04
    
    def __init__(self, first, last, pay):
        # instances
        self.first = first
        self.last = last
        self.pay = pay
        self.email = first + '.' + last + '@company.com'
                
        # Employee. instead of self. = constant class value
        Employee.num_of_emps += 1
        
    # method
    def fullname(self):
        return '{} {}'.format(self.first, self.last)
    
    def apply_raise(self):
        # self. instead of Employee = change the amount for single instance
        self.pay = int(self.pay * self.raise_amount)
        
    @classmethod
    # cls. is a class varaible name instead of a instance self.
    def set_raise_amount(cls, amount):
        cls.raise_amount = amount
    
```

```python
emp_1 = Employee('Corey', 'Schafer', 50000)
emp_2 = Employee('Test', 'User', 60000)

print(Employee.raise_amount)
print(emp_1.raise_amount)
print(emp_2.raise_amount)
```

```python
emp_1 = Employee('Corey', 'Schafer', 50000)
emp_2 = Employee('Test', 'User', 60000)

Employee.set_raise_amount(1.05)


print(Employee.raise_amount)
print(emp_1.raise_amount)
print(emp_2.raise_amount)
```

```python
class Employee:
    
    num_of_emps = 0
    raise_amount = 1.04
    
    def __init__(self, first, last, pay):
        # instances
        self.first = first
        self.last = last
        self.pay = pay
        self.email = first + '.' + last + '@company.com'
                
        # Employee. instead of self. = constant class value
        Employee.num_of_emps += 1
        
    # method
    def fullname(self):
        return '{} {}'.format(self.first, self.last)
    
    def apply_raise(self):
        # self. instead of Employee = change the amount for single instance
        self.pay = int(self.pay * self.raise_amount)
        
    @classmethod
    # cls. is a class varaible name instead of a instance self.
    def set_raise_amount(cls, amount):
        cls.raise_amount = amount
        
    @classmethod
    def from_string(cls, emp_str):
        first, last, pay = emp_str.split('-')
        return cls(first, last, pay) # return new employee object
        
```

### 3.2 Staticmethods

```python
class Employee:
    
    num_of_emps = 0
    raise_amount = 1.04
    
    def __init__(self, first, last, pay):
        # instances
        self.first = first
        self.last = last
        self.pay = pay
        self.email = first + '.' + last + '@company.com'
                
        # Employee. instead of self. = constant class value
        Employee.num_of_emps += 1
        
    # method
    def fullname(self):
        return '{} {}'.format(self.first, self.last)
    
    def apply_raise(self):
        # self. instead of Employee = change the amount for single instance
        self.pay = int(self.pay * self.raise_amount)
        
        
    # classmethods

    @classmethod
    # cls. is a class varaible name instead of a instance self.
    def set_raise_amount(cls, amount):
        cls.raise_amount = amount
        
    @classmethod
    def from_string(cls, emp_str):
        first, last, pay = emp_str.split('-')
        return cls(first, last, pay) # return new employee object
    
    # staticmethod
    # staticmethods don't take the instance or the class as the first argument
    @staticmethod # this is a decorator
    def is_workday(day):
        if day.weekday() == 5 or day.weekday() == 6: # saturday or sunday
            return False
        return True
```

```python
emp_1 = Employee('Corey', 'Schafer', 50000)
emp_2 = Employee('Test', 'User', 60000)

import datetime
my_date = datetime.date(2016, 7, 10)
print(Employee.is_workday(my_date))
```
