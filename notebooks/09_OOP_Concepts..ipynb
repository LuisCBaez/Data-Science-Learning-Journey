## Introduction to OOP

Object-Oriented Programming (OOP) is a programming paradigm based on the concept of objects, which can contain data (attributes) and behavior (methods). It helps organize code into reusable and modular components.

## Key Concepts of OOP

### Classes and Objects

- **Class**: A blueprint for creating objects.
- **Object**: An instance of a class.

```python
# Example
class Car:
    def __init__(self, make, model):
        self.make = make
        self.model = model

    def start_engine(self):
        print(f'The engine of {self.make} {self.model} has started.')

# Creating an object
my_car = Car('Toyota', 'Corolla')
my_car.start_engine()
```

### Attributes and Methods

Attributes are variables within a class. Methods are functions defined inside a class.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        print(f'Hello, my name is {self.name} and I am {self.age} years old.')

# Example usage
person = Person('Alice', 30)
person.greet()
```

### Encapsulation

Encapsulation restricts direct access to some of an object's components to protect data integrity.



```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # Private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance

# Example usage
account = BankAccount(1000)
account.deposit(500)
print(account.get_balance())
```

### Inheritance

Inheritance allows a class to inherit attributes and methods from another class.

```python
class Animal:
    def speak(self):
        print('Animal makes a sound')

class Dog(Animal):
    def speak(self):
        print('Dog barks')

# Example usage
dog = Dog()
dog.speak()
```

### Polymorphism

Polymorphism allows objects of different classes to be treated as objects of a common superclass.

```python
class Bird:
    def fly(self):
        print('Bird can fly')

class Penguin(Bird):
    def fly(self):
        print('Penguin cannot fly')

# Example usage
animals = [Bird(), Penguin()]
for animal in animals:
    animal.fly()
```

### Abstraction

Abstraction hides the complexity of certain operations and only shows essential features.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

# Example usage
rectangle = Rectangle(5, 10)
print(f'Area: {rectangle.area()}')
```

## Advanced OOP Concepts

### Magic (Dunder) Methods

Magic methods are special methods with double underscores that provide built-in functionality.

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f'({self.x}, {self.y})'

# Example usage
v1 = Vector(1, 2)
v2 = Vector(3, 4)
print(v1 + v2)
```

### Class and Static Methods


```python
class MathOperations:
    @staticmethod
    def add(a, b):
        return a + b

    @classmethod
    def describe_class(cls):
        return f'This is a class for math operations.'

# Example usage
print(MathOperations.add(5, 10))
print(MathOperations.describe_class())
```

### Property Decorators

```python
class Product:
    def __init__(self, price):
        self._price = price

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):
        if value > 0:
            self._price = value
        else:
            raise ValueError('Price must be positive')

# Example usage
product = Product(100)
product.price = 120
print(product.price)
```

## Design Patterns (Overview)

- **Singleton**: Ensures a class has only one instance.
- **Factory**: Creates objects without specifying the exact class.
- **Observer**: Notifies multiple objects when the state of one object changes.

## Example Project: Library Management System

Create a system to manage books, members, and borrowing.

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.is_available = True

class Member:
    def __init__(self, name):
        self.name = name
        self.borrowed_books = []

    def borrow_book(self, book):
        if book.is_available:
            book.is_available = False
            self.borrowed_books.append(book)
            print(f'{self.name} borrowed {book.title}')
        else:
            print(f'{book.title} is not available')

# Example usage
book1 = Book('1984', 'George Orwell')
member = Member('Alice')
member.borrow_book(book1)
```

## Best Practices in OOP

- Use meaningful names for classes, attributes, and methods.
- Keep methods short and focused.
- Apply the DRY (Don't Repeat Yourself) principle.
- Use composition over inheritance when appropriate.
- Follow Python's PEP 8 style guide.
