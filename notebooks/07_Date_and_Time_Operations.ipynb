## 1 Getting Started with the datetime Module
### 1.1 Importing the Module
First, import the module:

```python
import datetime
```

## 2  Creating and Displaying Datetime Objects

### 2.1 Current Date and Time
You can get the current local date and time using datetime.now()

### 2.2 Creating Specific Datetime Objects
Create datetime objects by providing the year, month, day (and optionally hour, minute, second)

```python
# Date and time now
now = datetime.datetime.now()
print(now)

# Flash crash - May 28, 1962
flash_crash = datetime.datetime(1962, 5, 28)
print(flash_crash)

# Black Monday - October 19, 1987
black_monday = datetime.datetime(1987, 10, 19)
print(black_monday)
```

## 3 Parsing Dates from Strings
Python’s datetime module allows you to convert date strings into datetime objects using the strptime() method. You provide a format string that tells Python how to parse the text.

### 3.1 Example: Mini-Crash of 1989
Consider the string date for a mini-crash event

```python
# Mini-crash of 1989 (example text)
crash_text = "Friday the 13th, Oct, 1989"

# Format codes used:
#   %A  - Full weekday name (e.g., "Friday")
#   %d  - Day of the month with a leading zero (e.g., "13")
#   %b  - Abbreviated month name (e.g., "Oct")
#   %Y  - Four-digit year (e.g., "1989")
# Note: The literal text (like " the " and commas) must match exactly.
crash_format_str = "%A the %dth, %b, %Y"
min_crash = datetime.datetime.strptime(crash_text, crash_format_str)
print("Mini-Crash Date:", min_crash)
```

## 3.2 Example: Recession of 1990
Another example with a different format:

```python
# Recession of 1990
recession_text = "07/03/90"
# Format codes:
#   %m - Month as a zero-padded decimal number (e.g., "07")
#   %d - Day of the month with a leading zero (e.g., "03")
#   %y - Two-digit year (e.g., "90")
recession_format_str = "%m/%d/%y"
nineties_rec = datetime.datetime.strptime(recession_text, recession_format_str)
print("Recession Date:", nineties_rec)
```

## 4 Formatting Datetime Objects into Strings
Once you have a datetime object, you can format it into a human-readable string using the strftime() method.

### 4.1 Converting Formats
Let’s start with an original date string, convert it to a datetime object, and then format it into a new style:

```python
# Original string date
org_text = "Sep 16 1992"
# Format string for the original text:
#   %b - Abbreviated month name (e.g., "Sep")
#   %d - Day of the month with a leading zero (e.g., "16")
#   %Y - Four-digit year (e.g., "1992")
org_format = "%b %d %Y"

# Create a datetime object from the original string
black_wednesday = datetime.datetime.strptime(org_text, org_format)
print("Black Wednesday (original):", black_wednesday)

# Now, format the datetime into a new string:
#   %A - Full weekday name (e.g., "Wednesday")
#   %B - Full month name (e.g., "September")
#   %d - Day of the month with a leading zero
#   %Y - Four-digit year
new_format = "%A, %B %d, %Y"
new_text = black_wednesday.strftime(new_format)
print("Black Wednesday (formatted):", new_text)
```

## 5 Accessing Datetime Attributes
Each datetime object has attributes that allow you to extract individual components such as the year, month, day, hour, minute, and second. This is useful when you want to filter or manipulate specific parts of a date.

### 5.1 Example: Tech Bubble Crash Date

```python
from datetime import datetime

# March 10, 2000 Tech Bubble Crash
tech_bubble = datetime(2000, 3, 10)

# Access the year, month, and day
yr = tech_bubble.year
mth = tech_bubble.month
day = tech_bubble.day

print(f"Tech Bubble Crash Date -> Year: {yr}, Month: {mth}, Day: {day}")
```

### 5.2. Additional Attributes from Current Date and Time

```python
# Get the current date and time
now = datetime.now()

print("Year:", now.year)
print("Month:", now.month)
print("Day:", now.day)
print("Hour:", now.hour)
print("Minute:", now.minute)
print("Second:", now.second)
```

## 6 Date Arithmetic with timedelta
Python’s timedelta objects represent durations and allow you to perform arithmetic on datetime objects (e.g., adding or subtracting days, weeks, etc.).

### 6.1 Calculating Relative Dates

```python
from datetime import timedelta

# TARP passed on October 3, 2008
tarp = datetime(2008, 10, 3)

# Seven days before TARP
week_before = tarp - timedelta(days=7)
print("One Week Before TARP:", week_before)

# One week after TARP
week_after = tarp + timedelta(weeks=1)
print("One Week After TARP:", week_after)

# One year after TARP (approximately 52 weeks)
year_after = tarp + timedelta(weeks=52)
print("Approximately One Year After TARP:", year_after)
```

### 6.2 Calculating the Difference Between Dates

```python
# Calculate the difference between two dates
difference = tarp - week_before
print("Difference between TARP and a week before:", difference)
```

## 7 Timezone-Aware Datetimes
Python’s standard library also supports timezone-aware datetimes. In Python 3.9+, you can use the built-in zoneinfo module. Here’s a quick example using timezone.utc:

```python
from datetime import datetime, timezone

# Current UTC time (timezone-aware)
utc_now = datetime.now(timezone.utc)
print("Current UTC Time:", utc_now)
```
