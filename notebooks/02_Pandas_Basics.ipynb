<h2 id="data-structures">1. Data Structures</h2>

- ```Series```:  One-dimensional labeled array capable of holding any data type (integers, strings, floats, etc.).
- ```DataFrame```: Two-dimensional labeled data structure with columns of potentially different types.

Pandas provides both index and data labels, making it a powerful tool for data manipulation and analysis.

The rows are called **index**.

### Series:

````python
# Padas Series
planets = ["Mercury", "Venus", "Earth", "Mars", "Jupiter", "Saturn", "Uranus", "Neptune"]
df_planets = pd.Series(data=planets)
df_planets
````

````python
# Set index values
planets = ["Mercury", "Venus", "Earth", "Mars", "Jupiter", "Saturn", "Uranus", "Neptune"]
position = [1, 2, 3, 4, 5, 6, 7, 8]
df_planets = pd.Series(data=position, index=planets)
df_planets
````


````python
# Select multiple rows and columns
df.loc[1:3, 'Artist':'Length']
````

#### Conditional Selection

| Function           | Description                        |
|--------------------|------------------------------------|
| DataFrame.eq()    | Element-wise equality              |
| DataFrame.ne()    | Element-wise inequality            |
| DataFrame.lt()    | Element-wise less than             |
| DataFrame.le()    | Element-wise less than or equal to |
| DataFrame.gt()    | Element-wise greater than          |
| DataFrame.ge()    | Element-wise greater than or equal to |
| DataFrame.all()   | True if all elements are True      |
| DataFrame.any()   | True if any element is True        |
| DataFrame.notnull()| True for non-null elements        |
| DataFrame.isnull()| True for null elements             |
| DataFrame.isin()  | Element-wise membership test      |


````python
# Create Pandas Series
s1 = pd.Series([1, 2, 3, 4, 5])
s2 = pd.Series([5, 4, 3, 2, 1])

# Element-wise comparison
print("Element-wise comparison (greater than):")
print(s1 > s2)

# Element-wise logical AND
print("\nElement-wise logical AND:")
print((s1 > 2) & (s2 < 4))

# Element-wise membership test
print("\nElement-wise membership test:")
print(s1.isin([1, 3, 5]))

# Element-wise is null
s3 = pd.Series([1, None, 3, None, 5])
print("\nElement-wise is null:")
print(s3.isnull())
````

````python
# Create a Pandas DataFrame
df_1 = pd.DataFrame({'A': [1, 2, 3, 4, 5],
                   'B': [5, 4, 3, 2, 1]})

# Element-wise equality
print("Element-wise equality:")
print(df_1.eq(3))

# Element-wise inequality
print("\nElement-wise inequality:")
print(df_1.ne(3))

# Element-wise less than
print("\nElement-wise less than:")
print(df_1.lt(3))

# Element-wise less than or equal to
print("\nElement-wise less than or equal to:")
print(df_1.le(3))

# Element-wise greater than
print("\nElement-wise greater than:")
print(df_1.gt(3))

# Element-wise greater than or equal to
print("\nElement-wise greater than or equal to:")
print(df_1.ge(3))
````

<h2 id="cleaning">5. Data Cleaning and Preprocessing</h2>

````python
# Rename columns with dictionaries
new_columns = {'Music Recording Sales (millions)': 'Recroding Sales (M)',
               'Claimed Sales (millions)': 'Claimed Sales (M)',
               'Released.1' : 'Date Released',
               'Released': 'Year Released'               
               }
df.rename(columns=new_columns, inplace=True)
df
````

````python
# Convert 'Released' columns to datetime data type
df['Date Released'] = pd.to_datetime(df['Date Released'], format='%d-%b-%y')
df.dtypes
````

````python
# Remove missing values
df.dropna(inplace=True)
df
````

The ```.dropna()``` remove all the rows with nan values.
If you want to remove missing values considering some colums us ```.dropna(subset=['column1', 'column2', ...])```

````python
# Remove a column
df.drop(columns=['Year Released'], inplace=True)
df
````
